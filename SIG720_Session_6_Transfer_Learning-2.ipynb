# Transfer Learning
* Transfer learning consists of taking features learned on one problem, and leveraging them on a new, similar problem.

 > For instance, features from a model that has learned to many identify different vehicles may be useful to kick-start a model meant to identify different types of cars.

* Transfer learning is usually done for tasks where your dataset has too little data to train a full-scale model from scratch.

## The typical transfer-learning workflow

### Method 1: Trained Weight Initialization
1. Instantiate a base model and load pre-trained weights into it.
2. Freeze all layers in the base model.
3. Create a new model on top of the output of one (or several) layers from the base model.
4. Train your new model on your new dataset.


### Method 2: Freezing Parameters

(Lightweight implementation: Freeze all parameters, then fine-tune the output layer only)

1. Instantiate a base model and load pre-trained weights into it.
2. Run your new dataset through it and record the output of one (or several) layers from the base model. This is called feature extraction.
3. Use that output as input data for a new, smaller model.

A key advantage of that second workflow is that you only run the base model once on your data, rather than once per epoch of training. So it's a lot faster & cheaper.

An issue with that second workflow, though, is that it doesn't allow you to dynamically modify the input data of your new model during training, which is required when doing data augmentation, for instance.

Transfer learning is typically used for tasks when your new dataset has too little data to train a full-scale model from scratch, and in such scenarios data augmentation is very important.

## Freezing Weights vs Freezing Parameters in Transfer Learning

| **Concept**             | **Freezing Weights**                                                                                     | **Freezing Parameters**                                                                                  |
|-------------------------|----------------------------------------------------------------------------------------------------------|----------------------------------------------------------------------------------------------------------|
| **Definition**          | Freezing weights means keeping the specific learned values of certain layers or parts of a neural network unchanged<br><br> during additional training.<br><br> This allows the model to retain previously learned features while enabling further training<br><br> or fine-tuning of other parts of the model to adapt to new data or tasks. | Freezing parameters means keeping all aspects of the model's configuration—<br><br>including weights, biases, and other settings—unchanged during further training.<br><br> This approach ensures that the entire model’s learned knowledge and structure remain static,<br><br> preventing any adaptation or learning from new data. |
| **How It Works**       | The model retains its previously acquired knowledge about specific features or patterns (from frozen layers)<br><br> and allows only the unfrozen layers to adapt to new information.<br><br> This selective training helps leverage existing learned features<br><br> while adjusting to new tasks or data. | The entire model's setup, including all weights and biases, remains fixed.<br><br> As a result, the model’s ability to learn or adapt is entirely halted,<br><br> meaning it does not adjust to any new data or changes in task requirements.<br><br> This approach maintains the original model's performance but limits its flexibility. |
| **Analogy**            | Imagine a chef who has mastered making pizza.<br><br> If they freeze the weights, it’s like they keep their pizza-making skills unchanged <br><br>while learning to prepare new dishes, like pasta, using only their existing knowledge <br><br>of ingredients and cooking techniques. | Consider a factory that makes toys with a set design and production process. <br><br>Freezing all parameters means the factory continues to produce toys exactly as it did before,<br><br> without any changes or improvements to the design or production methods,<br><br> even if new trends or technologies emerge. |
| **Example**            | **Image Classification**: Suppose you have a model trained to recognize cats and dogs.<br><br> If you freeze the weights of the initial layers,<br><br> the model will keep its knowledge of basic image features (like edges and textures)<br><br> while you adjust the final layers to recognize new breeds of dogs. | **Text Classification**: Imagine a model trained to detect spam emails with a set structure of weights and biases.<br><br> If you freeze all parameters, the model's spam detection ability remains fixed,<br><br> and it won't adapt to new types of spam or changes in spam tactics. |
| **Example**            | **Fashion MNIST**: Suppose you use a model pre-trained on MNIST digits.<br><br> If you freeze the weights of the initial layers and retrain only the last layers,<br><br> the model can still recognize fundamental patterns (like edges and textures)<br><br> while adjusting to classify different clothing items in the Fashion MNIST dataset. | **Traffic Signs**: If you have a model trained on recognizing traffic signs and you freeze all parameters,<br><br> the model’s ability to classify traffic signs remains static.<br><br> This means it won’t adapt to new or different signs that may appear over time. |


## Caution:
1. Transfer learning is not a silver bullet.

2. Which base model to use?
  * One size does not fit all.
  * If the base model is a binary classifier, but the new task is for multi-class classification, then transfer learning might fail.
  * Study the architecture of base model then apply. Don't start by first selecting the base model. In other words, the model should fit the data and the task, not the other way around.

3. Transfer Learning will only work when it makes sense.

# Why is transfer learning useful?
1. Stand on the shoulders of giants.

2. To leverage a fairly large well-trained model over a small dataset.

3. Fine-tune existing trained model to your unique data.

# When is Transfer Learning Appropriate?

1. Your problem is <u>similar</u> to a problem solved by someone else (and that model is also published).

2. The model was trained on a lot more data than what you have.

3. The base model is deep; shallow models are less likely to transfer well; unless the source and target data are very similar.


#### **2. Preprocess the Data**


```python
# Normalize and reshape data
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0
x_train = np.expand_dims(x_train, -1)
x_test = np.expand_dims(x_test, -1)

# Convert labels to one-hot encoding
y_train = tf.keras.utils.to_categorical(y_train, num_classes=10)
y_test = tf.keras.utils.to_categorical(y_test, num_classes=10)
```


#### **3. Define and Train LeNet-5 Model from Scratch**



## LeNet5 Architecture

![link text](https://d2l.ai/_images/lenet.svg)

```python
# Function to create the base model as per the LeNet-5 Architecture.
def create_lenet5_model():
    model = tf.keras.models.Sequential([
        tf.keras.layers.Input(shape=(28, 28, 1)),  # Define input shape with Input layer
        tf.keras.layers.Conv2D(6, (5, 5), activation='tanh', padding='same'),
        tf.keras.layers.AveragePooling2D(pool_size=(2, 2), strides=(2, 2)),
        tf.keras.layers.Conv2D(16, (5, 5), activation='tanh'),
        tf.keras.layers.AveragePooling2D(pool_size=(2, 2), strides=(2, 2)),
        tf.keras.layers.Conv2D(120, (5, 5), activation='tanh'),
        tf.keras.layers.Flatten(),
        tf.keras.layers.Dense(84, activation='tanh'),
        tf.keras.layers.Dense(10, activation='softmax')
    ])
    return model
```

```python
# Create, compile, and train the base model
base_model = create_lenet5_model()
base_model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
base_model.summary()
history = base_model.fit(x_train, y_train, epochs=5, batch_size=64, validation_split=0.1, verbose=2)
```

## View the base model

```python
tf.keras.utils.plot_model(base_model, show_shapes=True)
```

```python
# Define Function to plot the history of the model.
# Plot the accuracy and loss for train and validation sets.
def plot_training_history(history, model_name='model'):
    acc = history.history['accuracy']
    val_acc = history.history['val_accuracy']

    loss = history.history['loss']
    val_loss = history.history['val_loss']

    plt.figure(figsize=(8, 8))

    # Plotting accuracy
    plt.subplot(2, 1, 1)
    plt.plot(acc, label='Training Accuracy')
    plt.plot(val_acc, label='Validation Accuracy')
    plt.legend(loc='lower right')
    plt.ylabel('Accuracy')
    plt.ylim([min(min(acc), min(val_acc)), 1])
    plt.title(f'Training and Validation Accuracy for {model_name}')

    # Plotting loss
    plt.subplot(2, 1, 2)
    plt.plot(loss, label='Training Loss')
    plt.plot(val_loss, label='Validation Loss')
    plt.legend(loc='upper right')
    plt.ylabel('Cross Entropy')
    plt.ylim([0, 1.0])
    plt.title(f'Training and Validation Loss for {model_name}')
    plt.xlabel('Epoch')

    plt.tight_layout()  # Ensures labels do not overlap
    plt.show()
```

```python
# Plot the training history for the base model
plot_training_history(history = history, model_name='base_model')
```


#### **4. Define Transfer Learning with Pre-trained LeNet-5**



**a. Transfer Learning by Freezing Parameters**


```python
def create_transfer_model_freeze_params(base_model):

    #     Freeze Layers: By setting layer.trainable to False, you are freezing the layers, meaning their weights will not be updated during training.
    #     This is a common practice in transfer learning to retain the learned features of the pre-trained model while only training new layers added on top.

    for layer in base_model.layers:
        layer.trainable = False

    #     Initialize Sequential Model: models.Sequential() creates a new sequential model, which is a linear stack of layers.
    #     Add Pre-trained Model: base_model is added as the first layer in the new model. Since base_model has its layers frozen, it will only act as a feature extractor.

    #     Flatten Layer: layers.Flatten() is used to flatten the output of the base model (which is typically in the form of a multi-dimensional array)
    #     into a one-dimensional vector, which is needed before feeding into a Dense layer.

    #     Dense Layer: layers.Dense(10, activation='softmax') adds a dense (fully connected) layer with 10 units (one for each class)
    #     and a softmax activation function. The softmax function will output probabilities for each class.

    model = models.Sequential([
        base_model,
        layers.Flatten(),
        layers.Dense(10, activation='softmax')
    ])

    #     Compile Model: This configures the model for training.
    #     Optimizer: 'adam' is an optimization algorithm that adjusts weights based on gradients.
    #     Loss Function: 'categorical_crossentropy' is used for multi-class classification problems where the target variable is one-hot encoded.
    #     Metrics: ['accuracy'] specifies that accuracy should be monitored during training and evaluation.

    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    return model
```

```python

#     Create Transfer Model: This line creates a new model using the create_transfer_model_freeze_params function, with base_model as the input.
#     This new model will have the pre-trained layers frozen and additional layers added for the specific classification task.

model_transfer_params = create_transfer_model_freeze_params(base_model)
history_transfer_params = model_transfer_params.fit(x_train, y_train, epochs=10, validation_split=0.1, verbose=2)
```


**b. Transfer Learning by Freezing Weights**


```python
# In transfer learning, it's common to start by unfreezing all layers of a pre-trained model to ensure that you have the
# flexibility to fine-tune the entire network if necessary.
#  Then, specific layers are frozen again based on the desired training strategy. Here's why this approach is used.

def create_transfer_model_freeze_weights(base_model):

  #     Step 1:  Unfreeze all layers first

  #     Purpose: The goal here is to ensure that all layers of the pre-trained model are initially set to be trainable.
  #     This step ensures that you start with a clean slate, where no layers are unintentionally frozen.
  #     Reason: By setting all layers to trainable=True, you allow the model to be fine-tuned fully if needed.
  #     This can be useful in cases where the new task is quite different from the original task the base model was trained on.
  #             Example: The base model is trained to identify 100 categories but the new dataset has only 10 categories.

    for layer in base_model.layers:
        layer.trainable = True  # Unfreeze all layers first

  #    Step 2:  Freezing All But the Last Two Layers:
  #     Purpose: After unfreezing all layers, this step selectively freezes all but the last two layers of the base model.
  #     Reason: The idea is to keep the weights of the earlier layers frozen while allowing the last two layers (which are more task-specific) to be trained.
  #     The early layers of a neural network usually learn low-level features that are generally applicable across many tasks, such as edges or textures.
  #     In contrast, the later layers learn higher-level features that are more specific to the particular task.
  #     By freezing the early layers, you leverage the learned features without altering them, while allowing the last few layers to adapt to the new task.

    for layer in base_model.layers[:-2]:  # Freeze all but last two layers
        layer.trainable = False

  #     Why Unfreeze First: Starting with all layers unfrozen ensures that you have full control over which layers are trained.
  #     This approach can help in troubleshooting and ensures that no layer is inadvertently frozen.
  #     Why Freeze Again: After verifying that you have control over all layers, you can selectively freeze some layers to avoid overfitting and
  #     to leverage pre-trained features while adapting the model to the new task.

    model = models.Sequential([
        base_model,
        layers.Flatten(),
        layers.Dense(10, activation='softmax')
    ])

    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    return model
```

```python
# Create and train the transfer model
model_transfer_weights = create_transfer_model_freeze_weights(base_model)
history_transfer_weights = model_transfer_weights.fit(x_train, y_train, epochs=10, validation_split=0.1, verbose=2)
```


#### **5. Evaluate Models and Calculate Metrics**



| Metric                  | Description                                                                 | Formula                           | Possible Lowest Value   | Possible Highest Value  | Preferred Range          |
|-------------------------|-----------------------------------------------------------------------------|-----------------------------------|-------------------------|-------------------------|--------------------------|
| **Accuracy**            | Measures overall correctness of predictions.                                 | $\frac{TP + TN}{TP + TN + FP + FN}$ | 0.0                     | 1.0                     | 0.7 - 1.0                |
| **Precision**           | Reflects how many of the predicted positives were true positives.            | $\frac{TP}{TP + FP}$            | 0.0                     | 1.0                     | 0.7 - 1.0                |
| **Recall**              | Indicates the ability to identify all actual positives.                      | $\frac{TP}{TP + FN}$            | 0.0                     | 1.0                     | 0.7 - 1.0                |
| **F1 Score**            | Harmonic mean of precision and recall, providing balance between the two.    | $2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}$ | 0.0                     | 1.0                     | 0.7 - 1.0                |
| **ROC AUC Score**       | Evaluates the model's ability to distinguish between classes.                | Area under ROC curve              | 0.0                     | 1.0                     | 0.7 - 1.0                |
| **Youden’s Index**      | Combines sensitivity and specificity to evaluate overall test performance.   | $\text{Sensitivity} + \text{Specificity} - 1$ | -1.0                    | 1.0                     | 0.7 - 1.0                |
| **Test Loss**           | Measures the error of the model's predictions on the test set.               | Calculated by the model           | 0.0                     | Higher is worse          | -                        |
| **Test Accuracy**       | Measures the proportion of correctly classified instances in the test set.  | $\frac{\text{Number of correct predictions}}{\text{Total number of predictions}}$ | 0.0                     | 1.0                     | 0.7 - 1.0                |

**Define a method to evaluate a model against various metrics**

```python
def evaluate_model(model, x_test, y_test):
    # Predict probabilities and classes
    y_pred_prob = model.predict(x_test)
    y_pred_classes = np.argmax(y_pred_prob, axis=1)
    y_true = np.argmax(y_test, axis=1)

    # Classification Report
    report = classification_report(y_true, y_pred_classes, target_names=[str(i) for i in range(10)], output_dict=True)
    report_df = pd.DataFrame(report).transpose()

    # Confusion Matrix
    cm = confusion_matrix(y_true, y_pred_classes)
    cm_df = pd.DataFrame(cm, index=[str(i) for i in range(10)], columns=[str(i) for i in range(10)])

    # Accuracy
    accuracy = accuracy_score(y_true, y_pred_classes)

    # Precision, Recall, F1 Score
    precision = precision_score(y_true, y_pred_classes, average='weighted', zero_division=0)
    recall = recall_score(y_true, y_pred_classes, average='weighted', zero_division=0)
    f1 = f1_score(y_true, y_pred_classes, average='weighted', zero_division=0)

    # ROC AUC Score
    y_true_binary = tf.keras.utils.to_categorical(y_true, num_classes=10)
    roc_auc = roc_auc_score(y_true_binary, y_pred_prob, average='macro', multi_class='ovo')

    # Precision-Recall Curve (per class)
    precision_recall_curves = {}
    for i in range(10):
        precision_curve, recall_curve, _ = precision_recall_curve(y_true_binary[:, i], y_pred_prob[:, i])
        precision_recall_curves[f'Class {i}'] = pd.DataFrame({'Recall': recall_curve, 'Precision': precision_curve})

    # Youden's Index Calculation
    sensitivity = recall
    specificity = np.diag(cm) / np.sum(cm, axis=1)
    youden_index = sensitivity - (1 - specificity)

    # Handle cases where specificity is NaN (due to division by zero)
    youden_index = np.nan_to_num(youden_index)

    return {
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1': f1,
        'roc_auc': roc_auc,
        'youden_index': np.mean(youden_index),
        'report_df': report_df,
        'cm_df': cm_df,
        'precision_recall_curves': precision_recall_curves
    }
```

**a. Evaluate LeNet-5 from Scratch**



```python
metrics_scratch = evaluate_model(base_model, x_test, y_test)
```


**b. Evaluate Transfer Learning (Freezing Parameters)**


```python
metrics_transfer_params = evaluate_model(model_transfer_params, x_test, y_test)
```



**c. Evaluate Transfer Learning (Freezing Weights)**


```python
metrics_transfer_weights = evaluate_model(model_transfer_weights, x_test, y_test)
```

#### **6. Save and Plot Metrics**

**Confusion Matrix and Precision-Recall Curve**



```python
def plot_confusion_matrix(cm_df, filename, classes):

    plt.figure(figsize=(10, 7))
    tick_marks = np.arange(len(classes))
    plt.xticks(tick_marks, classes, rotation=45)
    plt.yticks(tick_marks, classes)

    sns.heatmap(cm_df, annot=True, fmt='d', cmap='Blues', xticklabels = classes, yticklabels=classes)

    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')

    plt.title('Confusion Matrix for ' + filename.replace('confusion_matrix_', '').strip('.png') + ' model')
    plt.savefig(filename)

    plt.tight_layout()
    plt.show()


# Save and plot for LeNet-5 from Scratch
plot_confusion_matrix(metrics_scratch['cm_df'], 'confusion_matrix_scratch.png', classes = class_names_fashion)

# Save and plot for Transfer Learning (Freezing Parameters)
plot_confusion_matrix(metrics_transfer_params['cm_df'], 'confusion_matrix_transfer_params.png', classes = class_names_fashion)

# Save and plot for Transfer Learning (Freezing Weights)
plot_confusion_matrix(metrics_transfer_weights['cm_df'], 'confusion_matrix_transfer_weights.png', classes = class_names_fashion)
```

<u>Let’s break down why the LeNet-5 architecture was selected for applying to the Fashion MNIST dataset in simple terms.</u>

### Why Use LeNet-5 for Fashion MNIST?

#### 1. **Historical Significance and Simplicity**
- **LeNet-5 is an Early CNN Model:** LeNet-5 was one of the first successful convolutional neural networks designed in the 1990s by Yann LeCun and his colleagues. It’s a classic model that has been widely studied and used in various applications. Its architecture, while simple compared to modern networks, is well-understood and serves as a good starting point for learning and experimentation.

#### 2. **Architecture Matches the Task**
- **Simple and Effective for Grayscale Images:** Fashion MNIST images are grayscale and relatively small (28x28 pixels). LeNet-5’s architecture is suitable for handling such small images. It uses a combination of convolutional layers and pooling layers to extract features from these images efficiently.

#### 3. **Feature Extraction Capabilities**
- **Convolutional Layers:** LeNet-5 uses convolutional layers to automatically detect features such as edges, textures, and patterns in images. These features are useful for recognizing different types of clothing in Fashion MNIST.

#### 4. **Proven Performance on Similar Datasets**
- **Works Well with MNIST-like Datasets:** LeNet-5 was originally designed for the MNIST dataset, which is also a small, grayscale dataset of handwritten digits. Since Fashion MNIST is similar in nature (grayscale, small images), LeNet-5 is a good choice for achieving effective results.

#### 5. **Training and Computational Efficiency**
- **Fewer Parameters:** LeNet-5 has fewer parameters compared to more complex modern architectures like VGG16 or ResNet. This means it’s computationally less demanding and faster to train, which is beneficial when working with smaller datasets like Fashion MNIST.

### Summary

- **LeNet-5’s Simplicity and Historical Success:** It’s an early yet effective CNN that works well for small grayscale images, making it a good fit for Fashion MNIST.
- **Feature Extraction for Clothing Classification:** The convolutional layers in LeNet-5 can effectively extract features needed for identifying different types of clothing.
- **Efficiency and Learning:** It’s less resource-intensive compared to more modern and complex models, making it suitable for a straightforward classification task like Fashion MNIST.



By starting with LeNet-5, you can understand the basics of convolutional neural networks and how feature extraction works, before moving on to more complex models for more challenging tasks.
